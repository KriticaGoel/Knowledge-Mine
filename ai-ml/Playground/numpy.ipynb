```python
sales = np.array([100, 200, 300, 400, 500, 600])
print(sales[0]) #100
print(sales[1:4]) #200 300 400
print(sales[:3]) #100 200 300
print(sales[::2]) #100 300 500
print(sales[-3:]) #400 500 600
```

```python
sales = np.array([
    [100, 200, 300],
    [400, 500, 600],
    [700, 800, 900]
])
print(sales[0, 0])
print(sales[1, :])
print(sales[:, 1])
print(sales[0:2, 1:3])
print(sales[:, 0])
```
